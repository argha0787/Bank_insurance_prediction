```python
import pandas as pd

df= pd.read_csv('../data/cleaned_dataset/cleaned.csv')
df.head()
```

```python
df.rename(columns= {'expenses': 'Medical Insurance Expenses'}, inplace= True)
```

```python
df.columns
```

```python
num_cols= df.select_dtypes(include= ['int', 'float']).columns
num_cols
```

```python
mean_comparision= df.groupby('Medical Insurance Expenses')[num_cols].mean()
mean_comparision
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
for col in df.columns:
    for col2 in df.columns:
        if col != col2 :
            plt.figure(figsize= (10, 5))
            plt.scatter(df[col], df[col2])
            plt.title(f'Comparision between {col} and {col2}')
            plt.xlabel(col)
            plt.ylabel(col2)
```

```python
df.to_csv('../data/data_analysed/ target_renamed.csv', index= False)
```
